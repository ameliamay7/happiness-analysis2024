The World Happiness Report is a landmark survey of the state of global happiness. The report continues to gain global recognition as governments, organizations, and civil society increasingly use happiness indicators to inform their policy-making decisions. Leading experts across fields – economics, psychology, survey analysis, national statistics, health, public policy and more – describe how measurements of well-being can be used effectively to assess the progress of nations. The reports review the state of happiness in the world today and show how the new science of happiness explains personal and national variations in happiness.

You may be asking why, besides curiosity, this is relevant? The WHR's information has been used in a variety of ways to understand and help the world

## Policy Guidance: 
Governments can use the results to design policies that improve citizens' well-being by addressing factors like healthcare, education, and economic stability. It provides actionable data on what contributes to happiness and how it can be improved at the societal level.

## Global Comparison:
It helps both us and policymakers understand how happiness varies around the world and highlights the cultural, economic, and political differences that impact quality of life. This can lead to better international cooperation on global issues such as poverty, inequality, and climate change.

## Public Awareness: 
By tracking happiness over time, the survey fosters public awareness about the importance of well-being in our daily lives and motivates individuals and communities to pursue policies and lifestyles that contribute to greater happiness.

## Holistic View of Progress: 
My personal most important use of the WHR, as traditional measures of national progress often focus on GDP or economic output. The World Happiness Survey challenges this narrow view and promotes a more complete understanding of progress by considering social and emotional factors, not just economic ones. At the end of the day, the most important part of a country is the people it serves. 



# Vocabulary to Know

## GDP per capita:
GDP per capita is a measure of a country's economic output that accounts for its number of people.
## Social support:
Social support means having friends and other people, including family, to turn to in times of need or crisis to give you a broader focus and positive self-image. Social support enhances quality of life and provides a buffer against adverse life events.
## Healthy life expectancy: 
Healthy Life Expectancy is the average number of years that a newborn can expect to live in "full health"—in other words, not hampered by disabling illnesses or injuries.
## Freedom to make life choices:
Freedom of choice describes an individual's opportunity and autonomy to perform an action selected from at least two available options, unconstrained by external parties.
## Generosity:  
The quality of being kind and generous.
## Perceptions of corruption: 
The Corruption Perceptions Index (CPI) is an index published annually by Transparency International since 1995 which ranks countries "by their perceived levels of public sector corruption, as determined by expert assessments and opinion surveys.

# Examining factors influencing international happiness


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df["Healthy life expectancy at birth"], y=df["Life Ladder"], alpha=0.6)
sns.regplot(x=df["Healthy life expectancy at birth"], y=df["Life Ladder"], scatter=False, color="red")
plt.title("Life Expectancy vs Happiness")
plt.xlabel("Healthy Life Expectancy at Birth")
plt.ylabel("Happiness Index (Life Ladder)")
plt.show()
```

We begin our exploration of the WHR by exploring some of the factors that the WHR evaluates as "happiness indicators/factors," such as country average life expentacy. Generally, we see a very strong relationship between Life Expectancy and Happiness. There are very few dots that fall out of the general pattern, with only around four countries strongly falling out of the pattern on the left, indicating that they have a very low healthy life expectancy, but still a happiness level on par with that of countries with 5-6x the life expentancy. However, 4 is not a high score on the life ladder/happiness metric, so it might not be out of the ordinary. Just as important, there are a few countries with high life expentacy but even lower happiness metrics than the countries with the lowest life expentancy, with two countries hovering below a score of 2. Countries with higher life expentancy are ofter more 'developed' nations with more advanced healthcare, meaning that life expentancy is not a vacuum, and instead a combination of many other factors. 

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df["Generosity"], y=df["Life Ladder"], alpha=0.6)
sns.regplot(x=df["Generosity"], y=df["Life Ladder"], scatter=False, color="red")
plt.title("Generosity vs Happiness")
plt.xlabel("Generosity")
plt.ylabel("Happiness Index (Life Ladder)")
plt.show()
```

Generosity is an extremely subjective perception and factor that the WHR uses, but it still brings up an interesting discussion in this graph. Overall, there is not a uniquely strong correlation between the two, especially with some unexpectedly low happiness ratings in high generosity countries. 

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df["Perceptions of corruption"], y=df["Life Ladder"], alpha=0.6)
sns.regplot(x=df["Perceptions of corruption"], y=df["Life Ladder"], scatter=False, color="red")
plt.title("Perception of Corruption vs Happiness")
plt.xlabel("Perceptions of Corruption")
plt.ylabel("Happiness Index (Life Ladder)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df["Log GDP per capita"], y=df["Social support"], hue=df["Life Ladder"], size=df["Life Ladder"], palette="coolwarm", alpha=0.7)
plt.title("Happiness and Economic Inequality")
plt.xlabel("Log GDP per Capita")
plt.ylabel("Social Support")
plt.legend(title="Happiness Index")
plt.show()
```

Overall, higher happiness levels (darker red dots) are generally associated with countries that have both higher log GDP per capita and higher social support values. There is a positive correlation between GDP per capita and social support, suggesting that better-off countries tend to also report stronger social networks, or at least perceptions of that social network through happiness. The top-right cluster of the graph shows the highest concentration of dark red dots (happiness index ~7.5), indicating that the happiest countries are both wealthy and socially supportive in a very concentrated and generally consistent way. There are almost no visible blue dots in the top right, although there are a few, suggesting that wealth - or wealth of a country - alone does not guarantee happiness. There are also very few countries with high happiness scores and low GDP, reinforcing the idea that economic prosperity is a significant driver, though not the only one. I would take from this the most important conclusion that economic and social well-being are both critical components of national happiness, with neither alone being sufficient for the highest levels of happiness.

# Generational and Regional Trends


Some things to note at an initial glance: Since 2006-10, happiness inequality has increased everywhere except Europe, with notable disparities by age group and gender. Negative emotions are rising worldwide except in East Asia and Europe, where they have decreased.



These next few graphs will examine these factors across regions. 

```python
import pandas as pd
import plotly.express as px

# Load the dataset
file_path = "DataForTable2.1.csv"
df = pd.read_csv(file_path)

# Clean year column (if it's not already int)
df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Drop rows with missing Life Ladder values
df_filtered = df.dropna(subset=["Life Ladder", "year", "Country name"])

# Create interactive line plot
fig = px.line(
    df_filtered,
    x="year",
    y="Life Ladder",
    color="Country name",
    title="Life Ladder (Happiness Score) Over Time by Country",
    labels={"year": "Year", "Life Ladder": "Happiness Score"},
)

# Add slider and selectors
fig.update_layout(
    xaxis=dict(
        rangeselector=dict(
            buttons=list([
                dict(count=5, label="Last 5 Years", step="year", stepmode="backward"),
                dict(count=10, label="Last 10 Years", step="year", stepmode="backward"),
                dict(step="all")
            ])
        ),
        rangeslider=dict(visible=True),
        type="linear"
    ),
    height=600,
    width=1000
)

fig.show()
```

Here, we can break down each country into a timeline of their happiness score. The easiest way to read this chart is through hovering over specific lines. While it can be hard to decipher, there are some major outliers, like how Finland has topped the charts since 2016 as well as the lowest happiness score by far (over 2 points) in Afghanistan. 

```python
import matplotlib.pyplot as plt
import seaborn as sns

east_asia = ['China', 'Japan', 'South Korea', 'Taiwan', 'Mongolia']
east_asia_df = df[df['Country name'].isin(east_asia)]

plt.figure(figsize=(10, 6))
sns.lineplot(data=east_asia_df, x='year', y='Life Ladder', hue='Country name', marker="o")
plt.title('Life Ladder Over Time in East Asia')
plt.ylabel('Average Happiness (Life Ladder)')
plt.xlabel('Year')
plt.grid(True)
plt.tight_layout()
plt.show()
```

That year marked strong economic recovery following the 2008 global financial crisis, with robust GDP growth and rising employment rates. The government also made substantial investments in education, healthcare, and technological infrastructure, contributing to overall quality of life. Additionally, national pride was high due to international recognition. 

```python
latin_america = ['Brazil', 'Mexico', 'Argentina', 'Colombia', 'Chile', 'Peru', 'Uruguay']
la_df = df[df['Country name'].isin(latin_america)]

plt.figure(figsize=(10, 6))
sns.lineplot(data=la_df, x='year', y='Negative affect', hue='Country name', marker="o")
plt.title('Negative Emotions in Latin America')
plt.ylabel('Average Negative Affect')
plt.xlabel('Year')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define regions manually
regions = {
    'Oceania': ['Australia', 'New Zealand'],
    'North America': ['United States', 'Canada'],
    'East Asia': ['China', 'Japan', 'South Korea'],
    'Europe': ['France', 'Germany', 'Sweden', 'Spain', 'Italy'],
    'South Asia': ['India', 'Pakistan', 'Bangladesh'],
}

# Assign region label
df['Region'] = df['Country name'].apply(lambda x: next((r for r, c in regions.items() if x in c), 'Other'))

# Filter and compute regional average Life Ladder per year
region_avg = df[df['Region'] != 'Other'].groupby(['year', 'Region'])['Life Ladder'].mean().reset_index()

# Plot
plt.figure(figsize=(10, 6))
sns.lineplot(data=region_avg, x='year', y='Life Ladder', hue='Region', marker='o')
plt.title('Average Happiness (Life Ladder) by Region Over Time')
plt.ylabel('Life Ladder (Happiness Score)')
plt.xlabel('Year')
plt.grid(True)
plt.tight_layout()
plt.show()

```

This graph is one of the best displays of how happiness changes across the globe depending on regions. It is also important to consider how some of these might be indicators of culture as well, and how people perceive happiness and themselves in different parts of the globe. Overall, Oceania and North America are home to the happiest people, while South Asia houses by far the least average happy people. 


```python
import pandas as pd
import plotly.express as px


# Clean and prepare data
df = df[['Country name', 'year', 'Life Ladder']]
df = df.dropna()

# Get available years
years = sorted(df['year'].unique())

# Create a figure with dropdown
fig = px.choropleth(
    df[df['year'] == years[0]],
    locations='Country name',
    locationmode='country names',
    color='Life Ladder',
    hover_name='Country name',
    color_continuous_scale='Viridis',
    title=f'Global Happiness (Life Ladder) in {years[0]}',
    range_color=(df['Life Ladder'].min(), df['Life Ladder'].max())
)

# Add dropdowns
dropdown_buttons = []

for year in years:
    visible_df = df[df['year'] == year]
    button = dict(
        method="restyle",
        args=[{
            'z': [visible_df['Life Ladder']],
            'locations': [visible_df['Country name']],
            'text': [visible_df['Country name']],
        },
        {'title': f'Global Happiness (Life Ladder) in {year}'}],
        label=str(year)
    )
    dropdown_buttons.append(button)

# Update layout
fig.update_layout(
    updatemenus=[{
        'active': 0,
        'buttons': dropdown_buttons,
        'x': 0.1,
        'y': 1.1,
        'xanchor': 'left',
        'yanchor': 'top'
    }],
    margin={"r":0,"t":50,"l":0,"b":0}
)

fig.show()
```

The same graph as before but as a life ladder by country on a map, allowing you to visually click through the years to see where happiness trends overlap. It also is a good display of when the WHR started to get enough data from most countries. 

```python
import pandas as pd
import plotly.graph_objects as go

# Define regions
regions = {
    'Oceania': ['Australia', 'New Zealand'],
    'North America': ['United States', 'Canada'],
    'East Asia': ['China', 'Japan', 'South Korea'],
    'Europe': ['France', 'Germany', 'Sweden', 'Spain', 'Italy'],
    'South Asia': ['India', 'Pakistan', 'Bangladesh'],
}

# Assign region labels
df['Region'] = df['Country name'].apply(lambda x: next((r for r, countries in regions.items() if x in countries), 'Other'))

# Define indicators to choose from
indicators = [
    'Life Ladder',
    'Social support',
    'Freedom to make life choices',
    'Generosity',
    'Perceptions of corruption',
    'Log GDP per capita',
    'Healthy life expectancy at birth'
]

# Precompute all region-averaged data
region_data = {}
for indicator in indicators:
    temp = df[df['Region'] != 'Other'].groupby(['year', 'Region'])[indicator].mean().reset_index()
    region_data[indicator] = temp

# Create base figure with the first indicator
initial_indicator = indicators[0]
fig = go.Figure()

for region in regions:
    subset = region_data[initial_indicator][region_data[initial_indicator]['Region'] == region]
    fig.add_trace(go.Scatter(
        x=subset['year'],
        y=subset[initial_indicator],
        mode='lines+markers',
        name=region,
        visible=True
    ))

# Add dropdown menu
dropdown_buttons = []
for i, indicator in enumerate(indicators):
    visibility = []
    for region in regions:
        visibility.extend([indicator == ind for ind in [indicator]*len(regions)])
    button = dict(
        label=indicator,
        method="update",
        args=[
            {"y": [region_data[indicator][region_data[indicator]['Region'] == region][indicator].values for region in regions],
             "x": [region_data[indicator][region_data[indicator]['Region'] == region]['year'].values for region in regions],
             "type": "scatter"},
            {"title": f"{indicator} by Region Over Time", "yaxis": {"title": indicator}}
        ]
    )
    dropdown_buttons.append(button)

# Update layout with dropdown
fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=dropdown_buttons,
            x=0.0,
            y=1.15,
            xanchor='left',
            yanchor='top'
        )
    ],
    title=f"{initial_indicator} by Region Over Time",
    xaxis_title="Year",
    yaxis_title=initial_indicator,
    template="plotly_white",
    height=600
)

fig.show()
```

The most conclusive graph of the regional set, this graph has a drop-down menu based on each world region, allowing users to examine a range of different factors like generosity, life ladder, and freedom to make life choices. While some of the graphs align well with others, many aren't what one would expect. Take some time to go through the dropdown menu, seeing what trends stick out to you the most.

# Limitations

It's easy to think of the WHI as a sure answer to the world's happiness meter. However, although much of the data makes sense and aligns with other studies, there are many limitations to consider in this field. First, the WHI uses a sample size of around 1,000 people per country for each survey year. To ensure a more precise estimate and reduce random sampling error, they often combine data from multiple years (e.g., 2012-2014 or 2022-2024) to achieve a larger sample size of at least 2,000 per country. However, this often dampens the effect of major events like war or famine that could happen in one year.

While the Global Happiness Index provides valuable insights, it has inherent biases and limitations. Some glaring issues are difficulty in finding causation with cross-sectional studies, survey limitations, and cultural differences, especially with the relatively small sample size. 


Last summer, I ran a market test survey, and I learned a lot about constructing a survey and the bias that goes into answering each question. The way you ask a question can easily make your respondents answer in one way or another. One example of this was ladder surveys, or sliding scales. Most people often don't want to put themselves at the extremes, instead somewhere in the middle to high range, if it is a positive attribute. Many are also just fine with putting themselves in the direct middle, because they think that the average or most people will converge to that number. 


Lastly, as I've been doing this project, I've been thinking a lot about how we think about happiness. Just because our lives are good doesn't necessarily mean we are happy, or believe ourselves to be happy. Happiness is instead a result of chemicals in our brain that are triggered by certain stimuli. Further, how we are able to reflect and process that happiness is also crucial to this study. My dopamine levels might be five times higher than the person next to me, and I wouldn't think on the outside that we are differently happy. But are we? When we broaden this to an international scale, we must start asking how culture and different societies impact the way we think about our own happiness. For instance, some countries inherently like to talk about bad news, while others, like Norway or Denmark, like to talk about how happy they are. The media and the echo chamber effect of what we perceive emotion to be changes how we perceive ourselves, especially in comparison to others. In the end, the ladder/scale survey that WHI uses requires you to think about being in a less happy situation, or less happy people. Beyond just response bias of people wanting to make themselves feel better about themselves by saying they are happier, since people often don't like to admit that they are sad, we must also consider how culture makes us think about our own emotions independently of others. “How can one reasonably conclude that country A is happier than country B, when happiness is being measured according to the way people in country A think about happiness?” asked the authors of a new study. On a more social level, throughout cultures and countries, perceptions of people's happiness are widely different due to their cultural evaluation of happiness.


Ultimately, WHI/R is a very solid start. But more broadly, the world might need to reconsider how we evaluate happiness. A more holistic approach, incorporating mental health, inequality, sustainability, and cultural factors, would provide a more accurate picture of global well-being.
